```python
##### Positional argument && Keyword Argumentss
#  >>>>>> Positional Argumentsss
def print_number(*numbee):
    for number in numbee:
        print(number)
print_number(1, 2, 3, 4, 5, "Rishiiiiiiiii")        
```

```python
def argu_number(* argu):
    for item in argu:
        print(item)
argu_number(1, 4, 6, 7, 5 ,  4, "Risi", "Alice", "Bob")
```
